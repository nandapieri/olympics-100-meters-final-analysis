## Data Source Validation

Before collecting the dataset, this test will verify that the selected video is accessible through the YouTube Data API v3 and that top-level comments are available. Video metadata and a small sample of up to 10 comments will help assess whether the source is suitable for subsequent collection. Comments are requested in relevance order, so this sample is not representative of the full discussion. Their language is not assumed to be English; language eligibility will be assessed in a later stage. The reported comment count, when available, is a video-level statistic and may include replies, whereas this test retrieves only top-level comments. This step performs no NLP analysis.


```
import requests

VIDEO_ID = "7Xnr805bm4E"
API_BASE = "https://www.googleapis.com/youtube/v3"


def youtube_get(resource, **params):
    """Return API data or a safe diagnostic without exposing request URLs."""
    try:
        response = requests.get(
            f"{API_BASE}/{resource}",
            params={**params, "key": YOUTUBE_API_KEY},
            timeout=30,
        )
    except requests.RequestException:
        print(f"{resource}: network request failed (connection or timeout).")
        return None

    try:
        data = response.json()
    except ValueError:
        print(f"{resource}: invalid JSON response (HTTP {response.status_code}).")
        return None

    if not response.ok or "error" in data:
        # Use structured reason codes only; raw errors may include credentials.
        reasons = [
            error.get("reason", "unknown")
            for error in data.get("error", {}).get("errors", [])
        ]
        if "commentsDisabled" in reasons:
            print("Comments are disabled for this video.")
        else:
            reason = ", ".join(reasons) or "unspecified API error"
            print(f"{resource}: HTTP {response.status_code}; reason: {reason}")
        return None
    return data


video_data = youtube_get("videos", part="snippet,statistics", id=VIDEO_ID)
if video_data is not None:
    videos = video_data.get("items", [])
    if not videos:
        print("No accessible video was returned for the supplied video ID.")
    else:
        video = videos[0]
        snippet = video.get("snippet", {})
        print("Title:", snippet.get("title", "Unavailable"))
        print("Channel:", snippet.get("channelTitle", "Unavailable"))
        print("Publication date:", snippet.get("publishedAt", "Unavailable"))
        print("Available comment count:", video.get("statistics", {}).get("commentCount", "Unavailable"))

        comment_data = youtube_get(
            "commentThreads",
            part="snippet",
            videoId=VIDEO_ID,
            maxResults=10,
            textFormat="plainText",
            order="relevance",
        )
        if comment_data is not None:
            comments = comment_data.get("items", [])[:10]
            print(f"Top-level comments returned: {len(comments)}")
            if not comments:
                print("No top-level comments were returned.")
            for number, item in enumerate(comments, start=1):
                text = item.get("snippet", {}).get("topLevelComment", {}).get("snippet", {}).get("textDisplay", "")
                preview = text if len(text) <= 150 else text[:147] + "..."
                print(f"{number}. {preview}")
```

## Data Collection

Top-level comments will be collected from the validated Olympic Games video, `7Xnr805bm4E`, using the YouTube Data API v3 and the existing API helper. Requests will retrieve pages of up to 100 comments in time order, following `nextPageToken` until 1,000 unique comments have been collected or no further pages are available. Duplicate comment IDs will be excluded. The CSV will retain each comment's ID, video ID, original plain-text content, publication timestamp, and like count; replies will not be collected.

Selection is limited to comments accessible through the API at collection time. No language filtering or text cleaning will be applied at this stage, so the dataset may contain multiple languages. Time ordering and the collection cap favour recently posted comments, while deleted, moderated, or otherwise unavailable comments cannot be included. The sample therefore does not represent all viewers or the complete historical discussion. API failures may result in a partial dataset, which will be identified in the collection output. Rerunning the cell will replace `comments.csv` with the new collection.


```
import csv
import json
from datetime import datetime, timezone

MAX_COMMENTS = 1000
fields = ["comment_id", "video_id", "text", "published_at", "like_count"]
collected_comments = []
seen_ids = set()
duplicate_count = 0
page_token = None
used_page_tokens = set()
collection_status = "Comment limit reached."
collection_started_at = datetime.now(timezone.utc).isoformat()

while len(collected_comments) < MAX_COMMENTS:
    params = {
        "part": "snippet",
        "videoId": VIDEO_ID,
        "maxResults": 100,
        "order": "time",
        "textFormat": "plainText",
    }
    if page_token:
        params["pageToken"] = page_token

    page = youtube_get("commentThreads", **params)
    if page is None:
        collection_status = "Stopped after an API/network error; any collected data are partial."
        break

    items = page.get("items", [])
    if not items:
        collection_status = "No further comments were returned."
        break

    for item in items:
        comment = item["snippet"]["topLevelComment"]
        comment_id = comment["id"]
        if comment_id in seen_ids:
            duplicate_count += 1
            continue
        seen_ids.add(comment_id)
        snippet = comment["snippet"]
        collected_comments.append({
            "comment_id": comment_id,
            "video_id": VIDEO_ID,
            "text": snippet["textDisplay"],
            "published_at": snippet["publishedAt"],
            "like_count": snippet.get("likeCount", 0),
        })
        if len(collected_comments) == MAX_COMMENTS:
            break

    if len(collected_comments) == MAX_COMMENTS:
        break
    page_token = page.get("nextPageToken")
    if not page_token:
        collection_status = "All available pages were retrieved."
        break
    if page_token in used_page_tokens:
        collection_status = "Stopped because the API repeated a page token; data may be partial."
        break
    used_page_tokens.add(page_token)

print("Collection started (UTC):", collection_started_at)
print("Collection status:", collection_status)
print("Unique top-level comments collected:", len(collected_comments))
print("Duplicate comment IDs excluded:", duplicate_count)
if collected_comments:
    dates = [datetime.fromisoformat(row["published_at"].replace("Z", "+00:00"))
             for row in collected_comments]
    print("Publication date range (UTC):", min(dates).isoformat(), "to", max(dates).isoformat())
    with open("comments.csv", "w", newline="", encoding="utf-8") as csv_file:
        writer = csv.DictWriter(csv_file, fieldnames=fields)
        writer.writeheader()
        writer.writerows(collected_comments)
    print("Saved to comments.csv")
    print("Preview of five records:")
    print(json.dumps(collected_comments[:5], ensure_ascii=False, indent=2))
else:
    print("Publication date range: unavailable (no comments collected).")
    print("No CSV written because no comments were collected.")
```

## Dataset Quality Assessment

Language distribution and text quality must be assessed before preprocessing because multilingual, missing, or low-content comments may affect the interpretation of subsequent frequency and sentiment results. This assessment reads the existing CSV without contacting YouTube and retains every record and its original text. Character and whitespace-delimited word lengths describe the raw comments; these are descriptive measures rather than linguistic tokenisation.

Language is estimated using the lightweight [langdetect library](https://pypi.org/project/langdetect/) (`langdetect==1.0.9`), with a fixed seed of 0. Short texts are particularly ambiguous. Comments with fewer than 20 alphabetic characters, failed detections, or a top detector score below 0.90 are labelled `unknown`. These pragmatic thresholds are not validated accuracy guarantees, and detector scores are not calibrated probabilities of correctness. Mixed-language comments may receive only one label. Manual inspection is therefore necessary before selecting English-language material.

Diagnostic flags identify timestamp-only text (one or more `m:ss` or `h:mm:ss` references), link-only text (`http://`, `https://`, or `www.`), and very little textual content (fewer than 10 alphabetic characters). Counts may overlap. Temporary removal of links for the last diagnostic count does not change stored text or the text submitted for detection. All assessment variables remain in memory; no comments are removed, cleaned, or written back to disk. The notebook environment requires pandas, matplotlib, and langdetect.


```
import re
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display
from langdetect import DetectorFactory, detect_langs, LangDetectException

DetectorFactory.seed = 0
comments_df = pd.read_csv("comments.csv", dtype={"comment_id": "string", "video_id": "string", "text": "string"})
quality_text = comments_df["text"].fillna("")
print("Dataset shape (rows, columns):", comments_df.shape)
print("Missing values by column:")
display(comments_df.isna().sum().rename("missing_values").to_frame())
print("Duplicate comment IDs (beyond first occurrence):", int(comments_df["comment_id"].dropna().duplicated().sum()))

lengths = pd.DataFrame({
    "characters": quality_text.str.len(),
    "whitespace_words": quality_text.str.split().str.len(),
})
print("Comment length distribution (non-missing text only):")
display(lengths.loc[comments_df["text"].notna()].astype(float).describe(percentiles=[0.25, 0.5, 0.75, 0.90, 0.95]).round(2))

link_pattern = r"(?:https?://|www\.)\S+"
time_pattern = r"(?:\d+:)?\d{1,3}:[0-5]\d"
timestamp_only = quality_text.str.fullmatch(rf"\s*{time_pattern}(?:\s+{time_pattern})*\s*")
link_only = quality_text.str.fullmatch(rf"\s*{link_pattern}(?:\s+{link_pattern})*\s*", flags=re.IGNORECASE)
very_little_text = quality_text.map(lambda text: sum(c.isalpha() for c in re.sub(link_pattern, "", text, flags=re.IGNORECASE)) < 10)
print("Low-content diagnostics (overlapping counts):")
display(pd.Series({
    "timestamp_only": int(timestamp_only.sum()),
    "link_only": int(link_only.sum()),
    "fewer_than_10_letters_excluding_links": int(very_little_text.sum()),
    "any_of_these_flags": int((timestamp_only | link_only | very_little_text).sum()),
}).rename("comments").to_frame())


def estimate_language(text):
    if sum(c.isalpha() for c in text) < 20:
        return "unknown"
    try:
        predictions = detect_langs(text)
        if predictions and predictions[0].prob >= 0.90:
            return predictions[0].lang
    except LangDetectException:
        pass
    return "unknown"


language_labels = quality_text.map(estimate_language)
language_distribution = language_labels.value_counts().rename_axis("language").to_frame("count")
language_distribution["percentage"] = (100 * language_distribution["count"] / len(comments_df)).round(1)
print("Estimated language distribution (ISO language codes):")
display(language_distribution)

fig, ax = plt.subplots(figsize=(10, 4))
language_distribution["count"].plot.bar(ax=ax, color="steelblue")
ax.set(title="Estimated language distribution of collected comments", xlabel="Language code / unknown", ylabel="Number of comments")
fig.tight_layout()
plt.show()

sample_groups = {
    "English": language_labels.eq("en"),
    "Non-English": ~language_labels.isin(["en", "unknown"]),
    "Unknown": language_labels.eq("unknown"),
}
for group, mask in sample_groups.items():
    print(f"{group}: {int(mask.sum())} comments; up to three reproducible examples")
    sample = comments_df.loc[mask, ["comment_id", "text"]].copy()
    sample["estimated_language"] = language_labels.loc[mask]
    if sample.empty:
        print("No comments in this category.")
    else:
        with pd.option_context("display.max_colwidth", None):
            display(sample.sample(n=min(3, len(sample)), random_state=0))
```

## Language Filtering and Sample Selection

Sample selection reuses `comments_df` and the initial `language_labels` without changing either. Timestamp-only and link-only comments are excluded first. A further exclusion applies to comments containing fewer than 10 alphabetic letters after URLs, timestamp references, and standalone numbers are removed from a temporary diagnostic string. This minimum is a methodological choice intended to retain interpretable language, not a validated boundary between useful and unusable comments.

Initially detected English comments meeting the content criterion are retained. Unknown comments are reassessed without the previous 20-letter gate. Recovery requires both an English detector score of at least 0.99 and at least two distinct words from an explicit English grammatical-cue list, including at least one of the stronger structural cues listed in the code. Matching is case-insensitive and uses whole words; Latin script, athlete names, or sporting terms alone do not qualify. The same rule applies to short and longer unknown comments. Previously detected non-English comments are excluded rather than relabelled.

This deliberately restrictive rule may miss valid English comments. Grammatical cues can occur in other languages or quotations, while detector scores are not calibrated accuracy estimates; recovered labels remain provisional, especially for mixed-language text. Samples are displayed for review, with no manual relabelling or translation. Diagnostic strings and word matches are used only for selection: `analysis_df` preserves the original text and metadata. Exclusion reasons are mutually exclusive and assigned in the order shown below, so each comment contributes to exactly one selection outcome. No preprocessing or additional substantive NLP analysis is performed.


```
# Run Dataset Quality Assessment first; no API calls are needed.
import re
import pandas as pd
from IPython.display import display
from langdetect import DetectorFactory, detect_langs, LangDetectException

DetectorFactory.seed = 0
selection_text = comments_df["text"].fillna("")
url_rule = r"(?:https?://|www\.)\S+"
timestamp_rule = r"(?<!\w)(?:\d+:)?\d{1,3}:[0-5]\d(?!\w)"


def diagnostic_content(text):
    value = re.sub(url_rule, " ", text, flags=re.IGNORECASE)
    value = re.sub(timestamp_rule, " ", value)
    return re.sub(r"\b\d+(?:[.,]\d+)*\b", " ", value)


content_for_selection = selection_text.map(diagnostic_content)
alphabetic_count = content_for_selection.map(lambda text: sum(c.isalpha() for c in text))
only_timestamps = selection_text.str.fullmatch(r"\s*(?:\d+:)?\d{1,3}:[0-5]\d(?:\s+(?:\d+:)?\d{1,3}:[0-5]\d)*\s*")
only_links = selection_text.str.fullmatch(rf"\s*{url_rule}(?:\s+{url_rule})*\s*", flags=re.IGNORECASE)

# Fixed linguistic cues, not labels assigned to individual comments.
english_cues = set("the this that what with without was were is are he his they you your it how but and not of to in for on a an".split())
structural_cues = set("the this that what with without was were are your they his how".split())


def recover_unknown_english(text):
    words = set(re.findall(r"\b[a-z]+\b", text.casefold()))
    if len(words & english_cues) < 2 or not (words & structural_cues):
        return False
    try:
        predictions = detect_langs(text)
        return bool(predictions and predictions[0].lang == "en" and predictions[0].prob >= 0.99)
    except LangDetectException:
        return False


selection_reason = pd.Series("excluded: unresolved language", index=comments_df.index, dtype="string")
recovered_english = pd.Series(False, index=comments_df.index)
for index in comments_df.index:
    if only_timestamps.loc[index]:
        selection_reason.loc[index] = "excluded: timestamp only"
    elif only_links.loc[index]:
        selection_reason.loc[index] = "excluded: link only"
    elif alphabetic_count.loc[index] < 10:
        selection_reason.loc[index] = "excluded: fewer than 10 letters"
    elif language_labels.loc[index] == "en":
        selection_reason.loc[index] = "retained: initially detected English"
    elif language_labels.loc[index] != "unknown":
        selection_reason.loc[index] = "excluded: initially detected non-English"
    elif recover_unknown_english(content_for_selection.loc[index]):
        recovered_english.loc[index] = True
        selection_reason.loc[index] = "retained: recovered unknown English"

retained_mask = selection_reason.str.startswith("retained:")
analysis_df = comments_df.loc[retained_mask].copy()
print("Original comments:", len(comments_df))
print("Retained:", len(analysis_df))
print("Excluded:", int((~retained_mask).sum()))
reason_order = [
    "retained: initially detected English",
    "retained: recovered unknown English",
    "excluded: timestamp only",
    "excluded: link only",
    "excluded: fewer than 10 letters",
    "excluded: initially detected non-English",
    "excluded: unresolved language",
]
selection_summary = selection_reason.value_counts().reindex(reason_order, fill_value=0).to_frame("comments")
selection_summary["percentage"] = (selection_summary["comments"] / len(comments_df) * 100).round(1)
display(selection_summary)

short_unknown = selection_text.map(lambda text: sum(c.isalpha() for c in text)).lt(20)
print("Recovered unknown comments with fewer than 20 original alphabetic letters:", int((recovered_english & short_unknown).sum()))
review_masks = {
    "Recovered short English comments": recovered_english & short_unknown,
    "Excluded comments (up to one example per reason)": ~retained_mask,
}
for title, mask in review_masks.items():
    print(title)
    review = comments_df.loc[mask, ["comment_id", "text"]].copy()
    review["initial_language"] = language_labels.loc[mask]
    review["selection_reason"] = selection_reason.loc[mask]
    if review.empty:
        print("No comments meet these criteria.")
    else:
        if title.startswith("Excluded"):
            review = review.groupby("selection_reason", sort=True).head(1)
        else:
            review = review.sample(n=min(5, len(review)), random_state=0)
        with pd.option_context("display.max_colwidth", None):
            display(review)

assert len(analysis_df) + int((~retained_mask).sum()) == len(comments_df)
assert analysis_df.equals(comments_df.loc[retained_mask])
```

## Text Preprocessing

Preprocessing is applied specifically to the selected comments for N-gram frequency analysis. A new `clean_text` column will contain the processed text, while the original `text` and metadata remain intact for later sentiment analysis, where punctuation, case, and emojis may convey meaning. No rows are removed, including comments that become empty.

Text is lowercased, and URLs, timestamp references, and standalone numbers are removed. Common negative contractions are expanded to preserve `not`; other common apostrophe endings are discarded before tokenisation. A regular expression extracts contiguous Unicode alphabetic words, excluding digits, punctuation, underscores, and emojis. Tokens are filtered using [scikit-learn's built-in English stopword list](https://github.com/scikit-learn/scikit-learn/blob/main/sklearn/feature_extraction/_stop_words.py), with `no`, `not`, `never`, `nor`, and `against` explicitly retained. Joining the remaining words with single spaces eliminates empty tokens and excess whitespace. The notebook environment requires scikit-learn.

No stemming or lemmatisation is used, preserving surface word forms and keeping the procedure transparent; related forms will therefore remain separate in later counts. Removing stopwords and punctuation can create adjacency between words that were not consecutive in the original text, so later N-grams must be interpreted as sequences in the processed representation rather than verbatim phrases.


```
import re
from sklearn.feature_extraction.text import ENGLISH_STOP_WORDS

retained_words = {"no", "not", "never", "nor", "against"}
ngram_stopwords = set(ENGLISH_STOP_WORDS) - retained_words


def preprocess_for_ngrams(text):
    text = text.lower()
    text = re.sub(r"(?:https?://|www\.)\S+", " ", text)
    text = re.sub(r"(?<!\w)(?:\d+:)?\d{1,3}:[0-5]\d(?!\w)", " ", text)
    text = re.sub(r"\b\d+(?:[.,]\d+)*\b", " ", text)
    # Preserve negation before punctuation is discarded.
    text = text.replace("’", "'")
    text = re.sub(r"\bcan't\b", "can not", text)
    text = re.sub(r"\bwon't\b", "will not", text)
    text = re.sub(r"\bshan't\b", "shall not", text)
    text = re.sub(r"n't\b", " not", text)
    text = re.sub(r"'(?:s|re|ve|ll|d|m)\b", " ", text)
    tokens = re.findall(r"[^\W\d_]+", text, flags=re.UNICODE)
    return " ".join(token for token in tokens if token and token not in ngram_stopwords)


original_analysis_text = analysis_df["text"].copy(deep=True)
analysis_df["clean_text"] = analysis_df["text"].fillna("").map(preprocess_for_ngrams)
empty_after_cleaning = analysis_df["clean_text"].eq("")
print("Comments retained (no rows removed):", len(analysis_df))
print("Comments empty after cleaning:", int(empty_after_cleaning.sum()))
print("Five original-versus-cleaned examples (fixed random seed):")
with pd.option_context("display.max_colwidth", None):
    display(analysis_df[["text", "clean_text"]].sample(n=min(5, len(analysis_df)), random_state=0))
assert analysis_df["text"].equals(original_analysis_text)
```

## N-gram Frequency Analysis

Bigrams and trigrams are sequences of two and three adjacent tokens, respectively. Their frequencies describe recurring word combinations within the selected English-language comments. This analysis uses the whitespace-separated tokens in `analysis_df["clean_text"]`, preserving their order within each comment. Separate `CountVectorizer` instances extract bigrams and trigrams using `tokenizer=str.split`, `token_pattern=None`, and `lowercase=False`. This configuration preserves the existing whitespace-separated tokens, including single-character tokens, without further preprocessing or stopword removal. Summing the document-term matrix columns counts all occurrences across comments, including repeated occurrences within a comment. No sequence crosses comment boundaries.

Because stopwords and punctuation were removed during preprocessing, some adjacent token pairs or triples were not consecutive in the original text. The outputs therefore describe combinations in the processed representation rather than necessarily verbatim phrases. The top 15 sequences of each length are presented in tables and horizontal bar charts. Ties are ordered alphabetically for reproducibility. These descriptive counts do not measure sentiment or establish the prevalence of opinions among all viewers.


```
from sklearn.feature_extraction.text import CountVectorizer
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display


def count_ngrams(texts, size):
    vectorizer = CountVectorizer(
        ngram_range=(size, size),
        tokenizer=str.split,
        token_pattern=None,
        lowercase=False,
        preprocessor=None,
        stop_words=None,
    )
    matrix = vectorizer.fit_transform(texts)
    frequencies = matrix.sum(axis=0).A1
    return {tuple(term.split()): int(frequency)
            for term, frequency in zip(vectorizer.get_feature_names_out(), frequencies)}


bigram_counts = count_ngrams(analysis_df["clean_text"], 2)
trigram_counts = count_ngrams(analysis_df["clean_text"], 3)


def top_ngram_table(counts, limit=15):
    ranked = sorted(counts.items(), key=lambda item: (-item[1], item[0]))[:limit]
    return pd.DataFrame([(" ".join(tokens), frequency) for tokens, frequency in ranked],
                        columns=["ngram", "frequency"])


top_bigrams = top_ngram_table(bigram_counts)
top_trigrams = top_ngram_table(trigram_counts)
print("Selected comments analysed:", len(analysis_df))
for label, table in [("Bigrams", top_bigrams), ("Trigrams", top_trigrams)]:
    print(f"Top 15 {label.lower()} (occurrence counts):")
    display(table)
    if table.empty:
        print(f"No {label.lower()} available.")
        continue
    fig, ax = plt.subplots(figsize=(10, 6))
    ax.barh(table["ngram"], table["frequency"], color="steelblue")
    ax.invert_yaxis()
    ax.set_title(f"Top 15 {label} in Selected YouTube Comments")
    ax.set_xlabel("Frequency (number of occurrences)")
    ax.set_ylabel(label[:-1])
    ax.xaxis.set_major_locator(plt.MaxNLocator(integer=True))
    fig.tight_layout()
    plt.show()
```

```
from wordcloud import WordCloud

top_100_bigrams = sorted(bigram_counts.items(), key=lambda item: (-item[1], item[0]))[:100]
bigram_phrase_frequencies = {" ".join(tokens): frequency
                             for tokens, frequency in top_100_bigrams}
bigram_wordcloud = WordCloud(
    width=1800,
    height=1100,
    background_color="white",
    color_func=lambda *args, **kwargs: "#245575",
    max_words=len(bigram_phrase_frequencies),
    max_font_size=162,
    min_font_size=6,
    margin=10,
    relative_scaling=1,
    prefer_horizontal=1.0,
    random_state=42,
    repeat=False,
).generate_from_frequencies(bigram_phrase_frequencies)

print("Distinct bigrams supplied:", len(bigram_phrase_frequencies))
print("Distinct bigrams placed:", len(bigram_wordcloud.layout_))
fig, ax = plt.subplots(figsize=(15, 9))
ax.imshow(bigram_wordcloud, interpolation="bilinear")
ax.set_title("Top 100 Bigrams in Selected YouTube Comments", fontsize=16)
ax.axis("off")
fig.tight_layout()
plt.show()
```

## Sentiment Analysis

VADER (Valence Aware Dictionary and sEntiment Reasoner) is a lexicon- and rule-based method developed for sentiment expressed in social media. Its handling of cues such as negation, intensifiers, capitalisation, and punctuation makes it a practical baseline for short English-language YouTube comments. This notebook uses [NLTK's SentimentIntensityAnalyzer](https://www.nltk.org/api/nltk.sentiment.vader.html) on the original `text`, preserving expressive information removed from `clean_text`. The environment requires NLTK; its `vader_lexicon` resource is downloaded only if unavailable.

The compound score aggregates lexical valence adjusted by VADER's rules and normalises the result to a scale from −1 to +1. It is not a probability or a measure of classification certainty. Following the [VADER scoring guidance](https://github.com/cjhutto/vaderSentiment#about-the-scoring), scores ≥ 0.05 are labelled positive, scores ≤ −0.05 negative, and intermediate scores neutral. The outputs describe automated classifications within the selected sample, not ground-truth labels or the reactions of all viewers.

Sarcasm, slang, sports-specific expressions, and contextual ambiguity may cause misclassification. A neutral label can also reflect unrecognised sentiment or offsetting language rather than a genuinely neutral opinion. Sentiment toward a comment's subject—for example, an athlete, a broadcaster, or another commenter—must not be interpreted as overall sentiment about the race. Three examples per category are selected with a fixed random seed for manual inspection; these examples do not constitute an accuracy evaluation.


```
import nltk
from nltk.sentiment import SentimentIntensityAnalyzer
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display

try:
    sentiment_analyzer = SentimentIntensityAnalyzer()
except LookupError:
    if not nltk.download("vader_lexicon", quiet=True):
        raise RuntimeError("VADER lexicon download failed; check network access and rerun this cell.")
    sentiment_analyzer = SentimentIntensityAnalyzer()

analysis_df["sentiment_score"] = analysis_df["text"].map(
    lambda text: sentiment_analyzer.polarity_scores(text)["compound"]
)


def label_sentiment(compound):
    if compound >= 0.05:
        return "positive"
    if compound <= -0.05:
        return "negative"
    return "neutral"


analysis_df["sentiment_label"] = analysis_df["sentiment_score"].map(label_sentiment)
sentiment_order = ["positive", "neutral", "negative"]
sentiment_distribution = analysis_df["sentiment_label"].value_counts().reindex(sentiment_order, fill_value=0).to_frame("count")
sentiment_distribution["percentage"] = (sentiment_distribution["count"] / len(analysis_df) * 100).round(2)
print("VADER sentiment classifications for original comments:")
display(sentiment_distribution)

fig, ax = plt.subplots(figsize=(7, 4))
ax.bar(sentiment_distribution.index, sentiment_distribution["count"], color=["seagreen", "slategray", "indianred"])
ax.set(title="VADER Sentiment Distribution in Selected Comments", xlabel="Sentiment category", ylabel="Number of comments")
ax.xaxis.set_tick_params(rotation=0)
ax.yaxis.set_major_locator(plt.MaxNLocator(integer=True))
fig.tight_layout()
plt.show()

for category in sentiment_order:
    print(f"{category.capitalize()} examples (fixed random seed):")
    category_rows = analysis_df.loc[analysis_df["sentiment_label"].eq(category), ["text", "sentiment_score", "sentiment_label"]]
    with pd.option_context("display.max_colwidth", None):
        display(category_rows.sample(n=min(3, len(category_rows)), random_state=0))
```

## Naive Bayes Sentiment Classification

This supervised sentiment baseline uses the public [UCI Sentiment Labelled Sentences dataset](https://archive.ics.uci.edu/dataset/331/sentiment+labelled+sentences), associated with Kotzias, Denil, de Freitas, and Smyth (2015), *From Group to Individual Labels Using Deep Features* ([paper](https://datalab.ics.uci.edu/papers/kdd2015_dimitris.pdf)). The authors manually labelled 1,000 English review sentences from each of Amazon, IMDb, and Yelp: 500 positive and 500 negative per source, totalling 3,000 examples. Original labels are `1` (positive) and `0` (negative). Clearly polar sentences were deliberately selected; there is no neutral class. This sentence-level annotated corpus is used instead of NLTK movie_reviews, whose labels are extracted from reviewers' ratings rather than independently assigned sentence-level annotations. Neither VADER outputs nor YouTube comments supply training labels.

A scikit-learn pipeline combines unigram `CountVectorizer` features with `MultinomialNB(alpha=1.0)`. Its default word tokenizer lowercases text and retains tokens of at least two word characters; no stopword removal is applied. This classifier-specific representation is separate from the existing N-gram preprocessing. A reproducible five-fold `StratifiedGroupKFold` split (seed 42) reserves the first fold for testing, giving approximately 80% training and 20% testing while balancing labels. Identical sentence texts are grouped so they cannot cross the train/test boundary. Vocabulary and classifier are fitted only on the training partition, with no tuning on the test set. Accuracy, per-class precision, recall, F1-score, macro/weighted averages, and a confusion matrix are reported on the held-out reference labels. The same training-partition model is then applied to the original YouTube texts.

The downloaded archive is read in memory without extracting files. Its checksum, actual counts, and split sizes are reported for reproducibility. Review sentences differ from sports comments in vocabulary, subject, and style. Deliberate class balance and exclusion of neutral examples also differ from the YouTube sample. Held-out results measure performance on this external review corpus, not accuracy on sports comments. Naive Bayes assumes conditional feature independence and has limited handling of word order, sarcasm, and contextual ambiguity.

The binary predictions are retained as positive or negative; no neutral labels are invented. Comparison with VADER's three categories is descriptive: matching/opposing polarity is examined only where VADER assigns positive or negative, and VADER-neutral comments are reported separately. Agreement is not evidence of correctness, and neither method provides ground-truth labels for YouTube. No existing text, preprocessing, VADER score, or label is changed.


```
import hashlib
import io
import zipfile
import requests
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.pipeline import make_pipeline
from sklearn.model_selection import StratifiedGroupKFold
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

training_url = "https://archive.ics.uci.edu/static/public/331/sentiment+labelled+sentences.zip"
try:
    training_response = requests.get(training_url, timeout=60)
    training_response.raise_for_status()
except requests.RequestException as error:
    raise RuntimeError("The external labelled dataset could not be downloaded; stop before training.") from None

training_rows = []
source_files = ["amazon_cells_labelled.txt", "imdb_labelled.txt", "yelp_labelled.txt"]
with zipfile.ZipFile(io.BytesIO(training_response.content)) as archive:
    for filename in source_files:
        member = next(name for name in archive.namelist() if name.endswith("/" + filename))
        for line in archive.read(member).decode("utf-8").split("\n"):
            if line.strip():
                sentence, label = line.rstrip("\r").rsplit("\t", 1)
                training_rows.append({"text": sentence, "label": {"0": "negative", "1": "positive"}[label], "source": filename})
nb_training_df = pd.DataFrame(training_rows)
print("Dataset: UCI Sentiment Labelled Sentences")
print("Archive SHA-256:", hashlib.sha256(training_response.content).hexdigest())
print("Total labelled examples:", len(nb_training_df))
print("Original label mapping: 0 = negative; 1 = positive")
display(pd.crosstab(nb_training_df["source"], nb_training_df["label"]))
print("Repeated sentence texts (beyond first occurrence):", int(nb_training_df["text"].duplicated().sum()))

nb_splitter = StratifiedGroupKFold(n_splits=5, shuffle=True, random_state=42)
nb_train_index, nb_test_index = next(nb_splitter.split(nb_training_df["text"], nb_training_df["label"], groups=nb_training_df["text"]))
nb_train = nb_training_df.iloc[nb_train_index]
nb_test = nb_training_df.iloc[nb_test_index]
assert set(nb_train["text"]).isdisjoint(set(nb_test["text"]))
print("Training examples:", len(nb_train), "| Held-out test examples:", len(nb_test))
display(pd.DataFrame({"train": nb_train["label"].value_counts(), "test": nb_test["label"].value_counts()}))

nb_classifier = make_pipeline(CountVectorizer(), MultinomialNB(alpha=1.0))
nb_classifier.fit(nb_train["text"], nb_train["label"])
nb_test_predictions = nb_classifier.predict(nb_test["text"])
nb_accuracy = accuracy_score(nb_test["label"], nb_test_predictions)
print(f"Held-out accuracy: {nb_accuracy:.4f}")
nb_report = pd.DataFrame(classification_report(nb_test["label"], nb_test_predictions, labels=["negative", "positive"], output_dict=True, zero_division=0)).T
print("Held-out classification report:")
display(nb_report.round(4))
nb_confusion = pd.DataFrame(confusion_matrix(nb_test["label"], nb_test_predictions, labels=["negative", "positive"]), index=pd.Index(["negative", "positive"], name="Reference label"), columns=pd.Index(["negative", "positive"], name="Predicted label"))
print("Held-out confusion matrix (rows = reference, columns = prediction):")
display(nb_confusion)
```

```
# Predictions are kept separately to preserve analysis_df and all existing labels.
nb_youtube_predictions = pd.Series(nb_classifier.predict(analysis_df["text"]), index=analysis_df.index, name="nb_label")
nb_comparison_df = analysis_df[["comment_id", "text", "sentiment_score", "sentiment_label"]].copy()
nb_comparison_df["nb_label"] = nb_youtube_predictions

comparison_distribution = pd.DataFrame(index=["positive", "neutral", "negative"])
comparison_distribution["VADER count"] = nb_comparison_df["sentiment_label"].value_counts().reindex(comparison_distribution.index, fill_value=0)
comparison_distribution["VADER %"] = (comparison_distribution["VADER count"] / len(nb_comparison_df) * 100).round(2)
comparison_distribution["Naive Bayes count"] = nb_youtube_predictions.value_counts().reindex(comparison_distribution.index).astype("Int64")
comparison_distribution["Naive Bayes %"] = (comparison_distribution["Naive Bayes count"] / len(nb_comparison_df) * 100).round(2)
print("YouTube label distributions (missing NB neutral values mean an unsupported class):")
display(comparison_distribution)
nb_vader_table = pd.crosstab(nb_comparison_df["sentiment_label"], nb_comparison_df["nb_label"]).reindex(index=["positive", "neutral", "negative"], columns=["positive", "negative"], fill_value=0)
print("Descriptive cross-tabulation; neither axis is ground truth:")
display(nb_vader_table)

vader_has_polarity = nb_comparison_df["sentiment_label"].ne("neutral")
matching_polarity = vader_has_polarity & nb_comparison_df["sentiment_label"].eq(nb_comparison_df["nb_label"])
opposing_polarity = vader_has_polarity & ~matching_polarity
print("Matching polarity among VADER-polar comments:", int(matching_polarity.sum()), "of", int(vader_has_polarity.sum()))
print("Opposing polarity:", int(opposing_polarity.sum()))
print("VADER-neutral comments (no equivalent NB category):", int((~vader_has_polarity).sum()))
for title, mask in [("Matching polarity", matching_polarity), ("Opposing polarity", opposing_polarity), ("VADER neutral / binary NB prediction", ~vader_has_polarity)]:
    print(title + " — up to three examples (seed=42):")
    examples = nb_comparison_df.loc[mask, ["text", "sentiment_score", "sentiment_label", "nb_label"]]
    with pd.option_context("display.max_colwidth", None, "display.max_columns", None):
        display(examples.sample(n=min(3, len(examples)), random_state=42))
```

### Naive Bayes Results and Comparison

The external corpus contained 3000 labelled sentences. The grouped stratified split used 2398 for training and 602 for testing. Held-out accuracy was 82.72%; macro precision, recall, and F1-score were 0.828, 0.828, and 0.827, respectively. These results apply to the reserved review sentences, not to YouTube sentiment accuracy.

Applied to the 618 original YouTube comments, the binary classifier predicted 224 positive (36.25%) and 394 negative (63.75%). VADER assigned 295 positive, 211 neutral, and 112 negative labels. Among the 407 comments with a polar VADER label, the methods assigned matching polarity to 216 and opposing polarity to 191. The 211 VADER-neutral comments received binary predictions because Naive Bayes has no neutral class; they are not counted as polarity disagreements.

Distributional differences and the displayed examples illustrate different model behaviour and label spaces, rather than establish which method is correct. Transferring a review-trained classifier to short sports comments introduces vocabulary and context differences, and the absence of neutral training examples forces binary decisions. No human-labelled YouTube test set is available, so this comparison cannot determine either method's accuracy on the target comments.


## Discussion

The research question concerned recurring word combinations and sentiment in the selected English-language comments. Of 1,000 collected top-level comments, 618 met the language and content criteria. Within this sample, **usain bolt** (27 occurrences) and **noah lyles** (23) were the leading processed bigrams. Other recurring combinations included **won race** (11), **kishane thompson** (10), **world champion** (10), and **gold medal** (9). These observed counts indicate recurring references to athletes, winning, and sporting achievement. Bolt's prominence is consistent with discussion extending beyond the finalists to comparisons with other athletes; the frequency alone does not establish the purpose or sentiment of those references.

Trigrams offered less stable material for interpretation: the highest frequency was only three occurrences, shared by **africa not usa**, **s speed km**, **speed km h**, and **th th th**. Some sequences contain fragments or units rather than meaningful phrases. Stopword removal can create adjacency absent from the original text, and retaining surface forms leaves spelling variants separate. Consequently, these outputs describe processed token sequences, not automatically coherent themes. The top-100 bigram word cloud offers an exploratory overview, while the bar charts provide more precise comparisons. Neither visualization measures the number of distinct people expressing an opinion.

VADER assigned 295 comments positive labels (47.73%), 211 neutral (34.14%), and 112 negative (18.12%). Positive was the largest category, but accounted for less than half of the sample. The illustrative 15-comment review highlights possible misclassifications and contextual ambiguity: **Lyles is an absolute legend 😂** received a neutral score of 0.0000 despite language that could express admiration; **Thomson will dominate soon** received a negative score of −0.1280 although it could express confidence in future performance. A comment combining an accusation of cheating with a claim of a Jamaican win received a positive score of 0.5859, showing why positive lexical cues need not imply approval of the result. These are interpretive concerns, not adjudicated errors or an accuracy estimate. Sarcasm, slang, emojis, and sports-specific meanings complicate interpretation, and a comment may target an athlete, officiating, or another subject rather than the race overall.

The Naive Bayes comparison reinforces this dependence on method and label space. VADER uses lexical valence and rules to assign three categories; the supervised CountVectorizer–MultinomialNB pipeline learned binary polarity from human-labelled review sentences. It predicted 224 positive comments (36.25%) and 394 negative (63.75%), with no neutral option. Among the 407 comments with polar VADER labels, 216 predictions matched and 191 opposed VADER's polarity; the 211 VADER-neutral comments were considered separately. These differences do not identify a superior classifier. Naive Bayes achieved 82.72% accuracy and macro F1 of 0.827 on its held-out UCI review data, but that evaluation does not establish accuracy on YouTube comments. Review vocabulary, deliberately polar training examples, and the model's limited representation of context constrain transfer to sports discussion.

The scope of inference is narrow. Comments came from one video and were retrieved non-randomly in time order under a 1,000-comment cap. Their publication dates span August 2024 to October 2026, so they cannot be treated as exclusively immediate reactions to the final. API availability and moderation may further affect inclusion. English-language filtering and the minimum-content rule excluded 382 comments; conservative recovery of short English comments reduced some losses but cannot eliminate language-detection errors or selection bias. The final sample therefore represents a defined subset of accessible discussion, not all public reactions. Preprocessing choices affect N-gram interpretation, while neither sentiment model has been evaluated against human reference labels for these YouTube comments.


## Conclusions

In answer to the research question, the selected comments were characterised by recurring athlete references, particularly **usain bolt** and **noah lyles**, alongside race-related combinations such as **won race** and **gold medal**. Trigram repetition was sparse and included sequences with limited semantic meaning. These findings support a descriptive account of recurring language rather than claims about dominant audience opinions.

VADER produced a distribution of 47.73% positive, 34.14% neutral, and 18.12% negative classifications. The qualitative examples and the contrasting binary Naive Bayes predictions demonstrate the need to distinguish automated polarity from context-specific interpretation. The external Naive Bayes evaluation supplies a reproducible review-domain benchmark, not validation of either method on sports comments. Together, the frequency and sentiment analyses answer the question for this selected sample while leaving the accuracy of target-domain sentiment labels unresolved. Broader claims would require more representative sampling and human evaluation of the YouTube classifications.
